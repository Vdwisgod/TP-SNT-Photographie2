## Partie A - Découverte du Module PIL pour créer une image en couleur ##

Le module PIL est une bibliothèque Python qui permet le traitement d'images numériques.

Dans un premier temps, on va voir qu'il permet  de créer des images en couleurs.

<b>1.</b> Le code ci-dessous permet de créer une image de définition 400 x 300 (400 colonnes et 300 lignes) dont les pixels sont rouges.
Lire le programme, puis l'exécuter.

```python
from PIL import Image    #charge le module PIL qui permet de manipuler les images avec python
largeur = 400
hauteur = 300

#Etape 1 Construire l’image

img = Image.new('RGB', (largeur, hauteur)) # On créé une image couleur 'RGB'. Par défaut, tous les pixels sont noirs

#Etape 2 : On parcourt tous les pixels, colonnes après colonnes

for i in range(largeur): # Balayage des colonnes
    for j in range(hauteur): # Balayage des lignes 
        
#Etape 3 : colorier chaque pixel de l'image
        img.putpixel((i, j), (255, 0, 0)) # Le pixel de coordonnées (i,j) est colorié en rouge

#Etape 4 : Afficher l’image et/ou l’enregistrer.
img.show() # la méthode show() permet de faire afficher l'image
img.save("rouge.png") # sauvegarde de l'image sous le nom rouge.png avec les fichiers annexes
```

<b>remarques sur l'étape 2</b> 

* Notre image contient 300 × 400 pixels. Impossible de les colorier un par un.
On utilise donc deux boucles imbriquées qui parcourent tous les pixels de l'image.
